```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
from scipy.linalg import expm

def resolver_mol_exponencial(N, T=1.0, alpha=1.0):
    h = 1.0 / (N + 1)
    x = np.linspace(0, 1, N + 2)

    u0 = np.sin(np.pi * x)
    u0_int = u0[1:-1]

    A = np.zeros((N, N), dtype=float)
    for i in range(0,len(A)):
      A[i,i]=-2.0
    for i in range(0,len(A)-1):
      A[i,i+1]=1.0
    for i in range(1,len(A)):
      A[i,i-1]=1.0
    A=(alpha/h**2)*A

    u_int = expm(A * T) @ u0_int
    u = np.concatenate(([0.0], u_int, [0.0]))

    return x, u

```

```python
x, u = resolver_mol_exponencial(25)
u_exacta = np.exp(-np.pi**2 * 1.0) * np.sin(np.pi * x)
plt.plot(x,u ,  label="Solucion numerica")
plt.plot(x,u_exacta,linestyle='--', color='#ff6A00', label="Solucion exacta")
plt.title("N=25, t=1")
plt.legend()
```
